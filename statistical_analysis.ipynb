```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

from scipy import stats
from scipy.stats import ttest_ind, f_oneway
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

sales_df = pd.read_csv("sales_data.csv")
churn_df = pd.read_csv("customer_churn.csv")

# Descriptive Statistics
print(sales_df.describe())
print(churn_df.describe())

# Histogram
sns.histplot(sales_df["Total_Sales"], kde=True)
plt.savefig("visualizations/histogram.png")
plt.show()

# Correlation Heatmap
corr = sales_df.select_dtypes(include="number").corr()

sns.heatmap(corr, annot=True)
plt.savefig("visualizations/heatmap.png")
plt.show()

# Hypothesis Test 1
group1 = sales_df[sales_df["Region"]=="North"]["Total_Sales"]
group2 = sales_df[sales_df["Region"]=="South"]["Total_Sales"]

print(ttest_ind(group1, group2))

# Hypothesis Test 2
yes = churn_df[churn_df["Churn"]=="Yes"]["MonthlyCharges"]
no = churn_df[churn_df["Churn"]=="No"]["MonthlyCharges"]

print(ttest_ind(yes, no))

# Hypothesis Test 3
groups = [g["Total_Sales"].values
          for _, g in sales_df.groupby("Region")]

print(f_oneway(*groups))

# Confidence Interval
mean = sales_df["Total_Sales"].mean()

ci = stats.t.interval(
    0.95,
    len(sales_df)-1,
    loc=mean,
    scale=stats.sem(sales_df["Total_Sales"])
)

print("95% CI:", ci)

# Regression
X = sales_df[["Quantity"]]
y = sales_df["Total_Sales"]

model = LinearRegression()
model.fit(X, y)

pred = model.predict(X)

print("R2:", r2_score(y, pred))

plt.scatter(X, y)
plt.plot(X, pred)
plt.savefig("visualizations/regression_plot.png")
plt.show()
```
